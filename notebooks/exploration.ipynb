````python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd()))

import pandas as pd

events = pd.read_parquet("data/raw/events.parquet")
features = pd.read_parquet("data/processed/user_features.parquet")
preds = pd.read_parquet("data/processed/predictions.parquet")
recommendations = pd.read_parquet("data/processed/recommendations.parquet")

events.head()
````

````python
print("Events:", len(events), "| Users:", events['user_id'].nunique())
print("Event types:")
events['event_type'].value_counts()
````

````python
# Predicted retention vs purchase propensity
preds.describe().T[['mean', 'min', 'max']]
````

````python
# Recommended action distribution
print(recommendations['action'].value_counts().to_string())
print("Mean confidence:", recommendations['confidence'].mean().round(3))
````
